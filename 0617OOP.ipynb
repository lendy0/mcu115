## Part 2：類別 (Class) 入門


# 類別的基本概念

**類別 (class)** 是物件的藍圖，定義物件的**屬性**（資料）與**方法**（行為）。

List、Tuple、Dict 都是類別，嚴格來說叫內建類別。
```python
## list
a = list()           # 建構子
b = []             # 字面量（常用）
c = list([1, 2, 3])   # 建構子 + 初始值
d = [1, 2, 3]      # 字面量（常用）

## tuple
a = tuple()          # 建構子
b = ()            # 字面量（常用）
c = tuple((1, 2, 3))    # 建構子 + 初始值
d = (1, 2, 3)        # 字面量（常用）

## dict
a = dict()           # 建構子
b = {}             # 字面量（常用）
c = dict({"key": "value"})   # 建構子 + 初始值
d = {"key": "value"}      # 字面量（常用）
e = dict(one=1, two=2)     # 建構子 + 關鍵字參數
f = dict([("one", 1), ("two", 2)])  # 建構子 + 串列
```

實務上字面量是首選，可讀性更好、寫法更簡潔。建構子形式通常只在需要轉換型別時才用，例如 list("abc") → ['a', 'b', 'c']。

## 自定義類別

```python
class Student:
    def __init__(self, name, age):  # __init__: 建構子，建立物件時自動呼叫
        self.name = name     # self: 代表物件本身
        self.age = age

    def greet(self):         # 方法
        return f"Hi, I'm {self.name}"

# 建立物件（實例化）
s = Student("Alice", 20)
print(s.greet())    # Hi, I'm Alice
print(s.name)       # Alice
```

| | 內建類別 (list/dict/tuple) | 自定義類別 (Student) |
|---|---|---|
| 建立方式 | 字面量 [] {} () | 一定呼叫 Student(...) |
| 建構子 | 有，但少用 | \_\_init\_\_ 定義 |
| 實例化 | 寫 x = [] 就是在實例化 | s = Student(...) 才實例化 |



**重點整理**
- `__init__`：初始化物件屬性的建構子，建立物件時自動執行
- `self`：第一個參數，指向物件本身
- **屬性**：`self.name` 這樣的變數
- **方法**：類別中定義的函式
- **實例化**：`Student(...)` = 建立一個物件

## Part 3：排序與進階類別概念


### `sorted()` 與 `lambda` 快速參考

#### `sorted()`
回傳排序後的新串列，不修改原資料。
```python
sorted(iterable, key=None, reverse=False)
sorted([3, 1, 2])             # [1, 2, 3]
sorted([3, 1, 2], reverse=True)      # [3, 2, 1]
sorted(["apple", "kiwi", "banana"], key=len)　# ['kiwi', 'apple', 'banana']
```

#### `lambda`
匿名函式，語法：`lambda 參數: 運算式`
```python
add = lambda x, y: x + y
print(add(3, 5))  # 8

# 常用於 sorted() 的 key 參數
data = [(1, "b"), (3, "a"), (2, "c")]
sorted(data, key=lambda x: x[1])      # [(3, 'a'), (1, 'b'), (2, 'c')]
```

#### 與 `.sort()` 的差別
- `sorted()`：回傳新串列，原資料不變
- `.sort()`：在原串列上直接排序

### 繼承 (Inheritance)
子類別繼承父類別的屬性與方法，可覆寫 (override) 或擴充。

```python
class Animal:
    def __init__(self, name):
        self.name = name
    def speak(self):
        return "..."

class Dog(Animal):
    def speak(self):
        return "汪汪"

class Cat(Animal):
    def speak(self):
        return "喵喵"

animals = [Dog("Buddy"), Cat("Kitty")]
for a in animals:
    print(f"{a.name} 說 {a.speak()}")
# Buddy 說 汪汪
# Kitty 說 喵喵
```

### 特殊方法 (Dunder Methods)
前後雙底線的方法，讓物件支援 Python 內建操作。

```python
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y
    def __str__(self):          # print() 時自動呼叫
        return f"({self.x}, {self.y})"
    def __add__(self, other):   # + 運算子
        return Point(self.x + other.x, self.y + other.y)

p1 = Point(1, 2)
p2 = Point(3, 4)
print(p1 + p2)  # (4, 6)
```

### 類別方法 vs 靜態方法

| | 實例方法 | `@classmethod` | `@staticmethod` |
|---|---|---|---|
| 第一個參數 | `self`（實例） | `cls`（類別） | 無 |
| 修飾器 | 不需要 | `@classmethod` | `@staticmethod` |
| 可存取實例屬性 | ✅ | ❌ | ❌ |
| 可存取類別屬性 | ✅ | ✅ | ❌ |

```python
class MyClass:
    count = 0
    @classmethod
    def increment(cls):
        cls.count += 1
    @staticmethod
    def util(n):
        return n * 2
```

### 封裝 (Encapsulation)
- `_name`：慣例為保護屬性，提示外部不應直接存取
- `__name`：私有屬性，Python 自動改寫為 `_ClassName__name`

```python
class Foo:
    def __init__(self):
        self._protected = 1    # 約定性保護
        self.__private = 2     # 名稱改寫為 _Foo__private
```

### 其他實用工具
- `isinstance(obj, Class)`：檢查物件是否為某類別或其子類別的實例
- `@property`：將方法變成屬性存取
