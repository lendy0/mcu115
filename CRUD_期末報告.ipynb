# Flask
Python Flask 是一種輕量級的網頁框架，只要少少數行程式碼，就可以架設網頁伺服器 (Web Server)。

## 步驟一:安裝 Flask 套件


## 步驟二:撰寫程式碼
```python
from flask import Flask

app = Flask(__name__)

@app.route("/")
def hello():
    return "<Title>lendy</Title><h1>hello world</h1>"

if __name__ == "__main__":
    app.run()
```

執行結果如下:


# Local develop (使用 localhost.run)
- development server (local) <- SSH Tunnel -> localhost.run server <- HTTP -> User (remote)

## localhost.run
不需要安裝任何額外套件，使用 SSH 內建功能即可建立公開 URL：

```bash
ssh -o StrictHostKeyChecking=no -R 80:localhost:5000 nokey@localhost.run
```

執行後會顯示類似：
```text
https://xxxx-xxxx-xxxx-xxxx-xxxx.localhost.run
```

將此 URL 貼到瀏覽器即可從外部存取 Flask 應用。

# 串接 Flask

```python
from flask import Flask, render_template, request, redirect, url_for, flash
import sqlite3 as sqlite

app = Flask(__name__)
app.secret_key = 'your-secret-key-here'  # 正式環境請改用 os.urandom(24)

@app.route("/")
@app.route("/index")
def index():
    with sqlite.connect("db_web.db") as con:
        con.row_factory = sqlite.Row
        cur = con.cursor()
        cur.execute("select * from users")
        data = cur.fetchall()
    return render_template("index.html", datas=data)

@app.route("/add_user", methods=['POST','GET'])
def add_user():
    if request.method == 'POST':
        try:
            uname = request.form['uname']
            contact = request.form['contact']
            with sqlite.connect("db_web.db") as con:
                cur = con.cursor()
                cur.execute("insert into users(UNAME,CONTACT) values (?,?)", (uname, contact))
            flash('User Added', 'success')
        except Exception as e:
            flash(f'Error: {e}', 'danger')
        return redirect(url_for("index"))
    return render_template("add_user.html")

@app.route("/edit_user/<string:uid>", methods=['POST','GET'])
def edit_user(uid):
    if request.method == 'POST':
        try:
            uname = request.form['uname']
            contact = request.form['contact']
            with sqlite.connect("db_web.db") as con:
                cur = con.cursor()
                cur.execute("update users set UNAME=?,CONTACT=? where UID=?", (uname, contact, uid))
            flash('User Updated', 'success')
        except Exception as e:
            flash(f'Error: {e}', 'danger')
        return redirect(url_for("index"))
    with sqlite.connect("db_web.db") as con:
        con.row_factory = sqlite.Row
        cur = con.cursor()
        cur.execute("select * from users where UID=?", (uid,))
        data = cur.fetchone()
    return render_template("edit_user.html", datas=data)

@app.route("/delete_user/<string:uid>", methods=['GET'])
def delete_user(uid):
    try:
        with sqlite.connect("db_web.db") as con:
            cur = con.cursor()
            cur.execute("delete from users where UID=?", (uid,))
        flash('User Deleted', 'warning')
    except Exception as e:
        flash(f'Error: {e}', 'danger')
    return redirect(url_for("index"))

if __name__ == '__main__':
    app.run(port=5000)
```

````python
# 寫入 Flask CRUD 應用檔並在背景執行
flask_crud_code = '''
from flask import Flask, render_template, request, redirect, url_for, flash
import sqlite3 as sqlite

app = Flask(__name__)
app.secret_key = 'your-secret-key-here'

@app.route("/")
@app.route("/index")
def index():
    with sqlite.connect("db_web.db") as con:
        con.row_factory = sqlite.Row
        cur = con.cursor()
        cur.execute("select * from users")
        data = cur.fetchall()
    return render_template("index.html", datas=data)

@app.route("/add_user", methods=['POST','GET'])
def add_user():
    if request.method == 'POST':
        try:
            uname = request.form['uname']
            contact = request.form['contact']
            with sqlite.connect("db_web.db") as con:
                cur = con.cursor()
                cur.execute("insert into users(UNAME,CONTACT) values (?,?)", (uname, contact))
            flash('User Added', 'success')
        except Exception as e:
            flash(f'Error: {e}', 'danger')
        return redirect(url_for("index"))
    return render_template("add_user.html")

@app.route("/edit_user/<string:uid>", methods=['POST','GET'])
def edit_user(uid):
    if request.method == 'POST':
        try:
            uname = request.form['uname']
            contact = request.form['contact']
            with sqlite.connect("db_web.db") as con:
                cur = con.cursor()
                cur.execute("update users set UNAME=?,CONTACT=? where UID=?", (uname, contact, uid))
            flash('User Updated', 'success')
        except Exception as e:
            flash(f'Error: {e}', 'danger')
        return redirect(url_for("index"))
    with sqlite.connect("db_web.db") as con:
        con.row_factory = sqlite.Row
        cur = con.cursor()
        cur.execute("select * from users where UID=?", (uid,))
        data = cur.fetchone()
    return render_template("edit_user.html", datas=data)

@app.route("/delete_user/<string:uid>", methods=['GET'])
def delete_user(uid):
    try:
        with sqlite.connect("db_web.db") as con:
            cur = con.cursor()
            cur.execute("delete from users where UID=?", (uid,))
        flash('User Deleted', 'warning')
    except Exception as e:
        flash(f'Error: {e}', 'danger')
    return redirect(url_for("index"))

if __name__ == '__main__':
    app.run(port=5000)
'''

with open("app_crud.py", "w", encoding="utf-8") as f:
    f.write(flask_crud_code)

import subprocess, sys
proc = subprocess.Popen([sys.executable, "app_crud.py"],
                       stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
import time
time.sleep(2)
print("Flask CRUD 已在背景啟動: http://127.0.0.1:5000")
print("=" * 50)
print("建立外部隧道指令:")
print("ssh -o StrictHostKeyChecking=no -R 80:localhost:5000 nokey@localhost.run")
````
