# 串接 Flask + CRUD
將上面的 CRUD 函式與 Flask 整合。


````python
from flask import request, redirect, url_for, flash

@app.route("/index")
@app.route("/")
def index_web():
    with sqlite.connect("db_web.db") as con:
        con.row_factory = sqlite.Row
        cur = con.cursor()
        cur.execute("select * from users")
        data = cur.fetchall()
    return render_template("index.html", datas=data)

@app.route("/add_user", methods=['POST','GET'])
def add_user_web():
    if request.method == 'POST':
        try:
            uname = request.form['uname']
            contact = request.form['contact']
            with sqlite.connect("db_web.db") as con:
                cur = con.cursor()
                cur.execute("insert into users(UNAME,CONTACT) values (?,?)", (uname, contact))
            flash('User Added', 'success')
        except Exception as e:
            flash(f'Error: {e}', 'danger')
        return redirect(url_for('index_web'))
    return render_template("add_user.html")

@app.route("/edit_user/<string:uid>", methods=['POST','GET'])
def edit_user_web(uid):
    if request.method == 'POST':
        try:
            uname = request.form['uname']
            contact = request.form['contact']
            with sqlite.connect("db_web.db") as con:
                cur = con.cursor()
                cur.execute("update users set UNAME=?,CONTACT=? where UID=?", (uname, contact, uid))
            flash('User Updated', 'success')
        except Exception as e:
            flash(f'Error: {e}', 'danger')
        return redirect(url_for('index_web'))
    with sqlite.connect("db_web.db") as con:
        con.row_factory = sqlite.Row
        cur = con.cursor()
        cur.execute("select * from users where UID=?", (uid,))
        data = cur.fetchone()
    return render_template("edit_user.html", datas=data)

@app.route("/delete_user/<string:uid>")
def delete_user_web(uid):
    try:
        with sqlite.connect("db_web.db") as con:
            cur = con.cursor()
            cur.execute("delete from users where UID=?", (uid,))
        flash('User Deleted', 'warning')
    except Exception as e:
        flash(f'Error: {e}', 'danger')
    return redirect(url_for('index_web'))

print("CRUD Web routes 已全部完成！")
print("執行最後一個 Cell 即可啟動 Server 並測試")
````

# 期末作業：學生管理系統

## 需求
1. 建立 **Student 類別**：學號(student_id)、姓名(name)、成績(score)、性別(gender)
2. **成績自動帶入學號後兩碼**
3. 使用 **SQLite** 管理全班學生
4. 使用 **Flask** 建立 Web 介面（新增、修改、刪除、查詢全部）
5. 最後使用 `serve_port()` + `app.run()` 啟動


````python
class Student:
    def __init__(self, student_id, name, gender):
        self.student_id = student_id
        self.name = name
        self.score = int(student_id[-2:])
        self.gender = gender

with sqlite.connect("student.db") as con:
    cur = con.cursor()
    cur.execute("DROP TABLE IF EXISTS students")
    cur.execute('''CREATE TABLE students (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        student_id TEXT NOT NULL,
        name TEXT NOT NULL,
        score INTEGER,
        gender TEXT
    )''')

print("Student 類別已定義，students 資料表已建立")
````

````python
def add_student(student_id, name, gender):
    score = int(student_id[-2:])
    with sqlite.connect("student.db") as con:
        cur = con.cursor()
        cur.execute("insert into students(student_id,name,score,gender) values (?,?,?,?)",
                   (student_id, name, score, gender))
        con.commit()
    print(f"已新增: {name}")

def list_students():
    with sqlite.connect("student.db") as con:
        con.row_factory = sqlite.Row
        cur = con.cursor()
        cur.execute("select * from students")
        return cur.fetchall()

def update_student(uid, student_id, name, gender):
    score = int(student_id[-2:])
    with sqlite.connect("student.db") as con:
        cur = con.cursor()
        cur.execute("update students set student_id=?, name=?, score=?, gender=? where id=?",
                   (student_id, name, score, gender, uid))
        con.commit()
    print(f"已修改 id={uid}")

def delete_student(uid):
    with sqlite.connect("student.db") as con:
        cur = con.cursor()
        cur.execute("delete from students where id=?", (uid,))
        con.commit()
    print(f"已刪除 id={uid}")

print("Student CRUD 函式已定義完成")
````

````python
from flask import Flask, render_template, request, redirect, url_for, flash
import os

app_student = Flask(__name__)
app_student.secret_key = 'student-secret-key'

os.makedirs("templates_student", exist_ok=True)
app_student.template_folder = "templates_student"

@app_student.route("/")
@app_student.route("/students")
def index_student():
    datas = list_students()
    return render_template("student_index.html", datas=datas)

@app_student.route("/students/add", methods=['POST','GET'])
def add_student_web():
    if request.method == 'POST':
        try:
            sid = request.form['student_id']
            name = request.form['name']
            gender = request.form['gender']
            add_student(sid, name, gender)
            flash('Student Added', 'success')
        except Exception as e:
            flash(f'Error: {e}', 'danger')
        return redirect(url_for('index_student'))
    return render_template("student_add.html")

@app_student.route("/students/edit/<int:uid>", methods=['POST','GET'])
def edit_student_web(uid):
    if request.method == 'POST':
        try:
            sid = request.form['student_id']
            name = request.form['name']
            gender = request.form['gender']
            update_student(uid, sid, name, gender)
            flash('Student Updated', 'success')
        except Exception as e:
            flash(f'Error: {e}', 'danger')
        return redirect(url_for('index_student'))
    with sqlite.connect("student.db") as con:
        con.row_factory = sqlite.Row
        cur = con.cursor()
        cur.execute("select * from students where id=?", (uid,))
        data = cur.fetchone()
    return render_template("student_edit.html", datas=data)

@app_student.route("/students/delete/<int:uid>")
def delete_student_web(uid):
    try:
        delete_student(uid)
        flash('Student Deleted', 'warning')
    except Exception as e:
        flash(f'Error: {e}', 'danger')
    return redirect(url_for('index_student'))

print("Student Flask routes 已定義完成！")
print("最後一步：執行下方 Cell 建立模板 + 啟動 Server")
````
